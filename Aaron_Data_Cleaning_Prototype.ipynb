```

import pandas as pd
import numpy as np

df = pd.read_csv("ai_impact_jobs_2010_2025.csv")

# Remove duplicates
df = df.drop_duplicates()

# Impute salary by industry median
df["salary_usd"] = df.groupby("industry")["salary_usd"].transform(lambda x: x.fillna(x.median()))
df["salary_usd"] = df["salary_usd"].fillna(df["salary_usd"].median())

# Impute ai_skills
df["ai_skills"] = df["ai_skills"].fillna("Not Specified")

# Remove outliers using IQR
Q1 = df["salary_usd"].quantile(0.25)
Q3 = df["salary_usd"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df["salary_usd"] >= lower_bound) & (df["salary_usd"] <= upper_bound)]

# Keep top 5 industries
top_sectors = df["industry"].value_counts().head(5).index
df = df[df["industry"].isin(top_sectors)]

df.to_csv("ai_impact_jobs_2010_2025_GOLD.csv", index=False)
```
